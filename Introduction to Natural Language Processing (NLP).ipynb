# Introduction
<img src="https://www.oriresults.com/wp-content/uploads/Blog-Whats-Hiding-in-Your-Unstructured-Data-1000x592px.png" alt="Unstructured Data" width="1200" height="600">

Di era digital saat ini, jumlah data yang dihasilkan sangat besar dan beragam. Salah satu bentuk data yang paling umum adalah data teks. **Data teks** mencakup segala jenis informasi yang ditulis dalam bentuk kata-kata, kalimat, atau paragraf. Contoh data teks meliputi artikel berita, postingan media sosial, email, dokumen, dan transkrip percakapan.

**Data tidak terstruktur (Unstructured Data)** adalah jenis data yang tidak memiliki format atau struktur yang tetap, seperti tabel yang terdiri atas baris dan kolom. Contoh data tidak terstruktur adalah data teks, gambar, video, dan audio. Karakteristik utama dari data tidak terstruktur adalah ketidakmampuannya untuk dengan mudah dimasukkan ke dalam model data tradisional tanpa proses pengolahan terlebih dahulu.

Data tidak terstruktur, khususnya teks, sangat berharga namun menimbulkan berbagai tantangan:
- Volume Besar: Jumlah data yang sangat besar membuat analisis manual menjadi tidak praktis.
- Variasi Bahasa: Adanya berbagai bahasa, dialek, slang, dan terminologi khusus yang berbeda.
- Konteks dan Ambiguitas: Bahasa manusia seringkali mengandung konteks yang kompleks dan ambiguitas yang sulit dipahami oleh mesin.

Untuk mengatasi tantangan yang terkait dengan data tidak terstruktur, terutama dalam bentuk teks, diperlukan teknologi yang mampu memahami dan menginterpretasikan bahasa manusia. Di sinilah **Natural Language Processing (NLP)** berperan penting.

### **WHAT IS NLP**?
<img src="https://lh5.googleusercontent.com/3-Mzg8FVXSvhQfHK_xH855vAHX4EgjG-hh8Km8F9bCf26paCi9k2CsTVlxtgsuJ99VdPEC2-WR1lGKO30QAyiaL-zr71-qdUraJsbO1sPoKJoKUD2LaG9vEpk_GTP7CTJJIKGWV89TsYIh-ssihtDVc" alt="NLP" width="1000" height="600">

**Natural Language Processing (NLP)** adalah cabang ilmu kecerdasan buatan (AI) yang berfokus pada pengolahan bahasa alami. Bahasa alami adalah bahasa yang umum digunakan oleh manusia untuk berkomunikasi. Agar komputer dapat memahami dan memproses bahasa ini, perlu dilakukan pengolahan terlebih dahulu agar maksud dari pengguna bisa dipahami dengan baik oleh komputer.
<br><br>
Ada berbagai aplikasi yang memanfaatkan NLP, yang paling populer adalah:
1. Question Answering Systems: Kemampuan komputer untuk menjawab pertanyaan yang diberikan oleh user. Contohnya adalah chatbot seperti Siri, Alexa, dan Google Assistant.
2. Summarization: Pembuatan ringkasan dari sekumpulan konten dokumen atau email. Dengan menggunakan aplikasi ini, user bisa dibantu untuk mengkonversikan dokumen teks yang besar ke dalam bentuk slide presentasi.
3. Machine Translation. Produk yang dihasilkan adalah aplikasi yang dapat memahami bahasa manusia dan menterjemahkannya ke dalam bahasa lain. Termasuk di dalamnya adalah Google Translate yang apabila dicermati semakin membaik dalam penterjemahan bahasa
4. Speech Recognition. Field ini merupakan cabang ilmu NLP yang cukup sulit. Proses pembangunan model untuk digunakan telpon/komputer dalam mengenali bahasa yang diucapkan sudah banyak dikerjakan. Bahasa yang sering digunakan adalah berupa pertanyaan dan perintah.
5. Document classification. Sedangkan aplikasi ini adalah merupakan area penelitian NLP Yang paling sukses. Pekerjaan yang dilakukan aplikasi ini adalah menentukan dimana tempat terbaik dokumen yang baru diinputkan ke dalam sistem. Hal ini sangat berguna pada aplikasi spam filtering, news article classification, dan movie review.









**NLP Pipeline**

<img src="https://miro.medium.com/v2/resize:fit:966/1*rJQVqDjbhI3k22lHqa4dFw.png" alt="NLP Pipeline" width="700" height="450">

1. Langkah pertama dalam NLP adalah **Data Acquisiton** atau mengumpulkan/memperoleh data yang relevan dengan tujuan yang ingin dicapai. Cara memperoleh data text dapat dilakukan dengan mengunduh dataset yang dapat digunakan secara public dari website tertentu atau dapat juga melakukan scrapping/crawling data dengan mengekstrak informasi dari website.

2. Data yang kita dapatkan jarang bersih, sehingga diperlukan tahapan **Text Cleaning** atau pembersihan teks. Pada tahapan ini kita biasanya melakukan beberapa tahapan seperti case folding atau proses mengubah semua karakter huruf dalam teks menjadi huruf kecil atau huruf besar, menghilangkan tanda baca, menghapus whitespace, menghapus/mentransformasi angka, menghapus stop word, dan text normalization/noise removal.

3. Setelah pembersihan, data teks seringkali memiliki banyak variasi dan perlu diubah menjadi bentuk kanonik (prinsip atau cara standar yang telah ditentukan sebelumnya). Misalnya mengubah bahasa tidak baku menjadi bahasa baku. Hal ini dilakukan dalam tahap **Pre-Processing**.

4. Berikutnya dilanjutkan dengan tahapan **Feature Engineering**, di mana kita mengidentifikasi indikator yang paling cocok untuk tugas yang sedang dihadapi. Tahapan ini dilakukan untuk mentransformasi data-data yang masih dalam bentuk text ke bentuk numerik, yang sering disebut Feature Extraction. Indikator/fitur-fitur ini diubah ke dalam format yang dapat dimengerti oleh algoritma pemodelan.

5. Selanjutnya masuk ke tahapan **Modeling and Evaluation**, di mana kita membangun satu atau lebih model dan membandingkan mereka menggunakan metrik evaluasi yang relevan.

6. Setelah model terbaik di antara yang dievaluasi dipilih, kita beralih ke tahap **Deployment** atau penyaluran model ke dalam project yang ingin dilakukan. Pada tahapan akhir, kita secara teratur melakukan **Monitoring and Model Updating** untuk memantau kinerja model tersebut, dan jika perlu, memperbarui untuk menjaga kinerjanya

## Text Cleaning

```python
def clean_text(text):
    # Menghapus username
    text = re.sub(r'@\w+', '', text)

    # Menghapus tag
    text = re.sub(r'#\w+', '', text)

    # Hapus teks dalam []
    text = re.sub(r'\[\w+]', '', text)

    # Penghapusan Karakter Non-Alfanumerik
    text = re.sub(r'[^\w\s]', '', text)

    # Menghapus URL
    text = re.sub(r'http\S+|www\S+', '', text)
    text = re.sub(r'\b\w+(?:com|org)\b', '', text)

    # Numbers Removal: Menghapus karakter numerik
    text = re.sub(r'\d+', '', text)

    # Case Folding: Menyeragamkan teks (lower case)
    text = text.lower()

    # Menghapus whitespace (spasi ganda, tab, newline)
    text = re.sub(r'\s+', ' ', text)
    text = text.strip()

    #  Punctuation Removal: Menghapus tanda baca (seperti titik, koma, tanda seru, tanda tanya, dan titik koma)
    text = text.translate(str.maketrans('', '', string.punctuation))

    # tambahkan metode cleaning lainnya

    return text
```

## Text Pre-Processing
Tahapan ini mencakup proses:
1. **Slang Words Handling**:
Slang Words Handling melibatkan penggantian kata-kata slang dengan kata-kata standar yang sesuai. Contohnya, mengganti "gue" dengan "saya". Langkah ini membantu memastikan bahwa teks yang diproses lebih mudah dipahami dan lebih relevan dalam analisis.

2. **Stemming**:
Stemming adalah proses menghilangkan infleksi kata ke bentuk dasarnya, namun bentuk dasar tersebut tidak berarti sama dengan akar kata (root word). Misalnya kata “mendengarkan”, “dengarkan”, “didengarkan” akan ditransformasi menjadi kata “dengar”.

3. **Stop Words Removal**:
Stop words adalah kata-kata umum yang sering muncul dalam teks tetapi tidak memberikan kontribusi signifikan terhadap makna teks, seperti "yang", "di", "dan", dll. Menghapus stop words membantu mengurangi noise dalam teks dan meningkatkan relevansi kata-kata yang tersisa dalam analisis.

```python
# Fungsi menghapus stopword
def stopwords_removal(words,stopword):
    return [word for word in words if word not in stopword]

def preprocess_text(text, slang_words = sl_word, sw = sw_id):
    # SlangWords Handling
    text = ' '.join([slang_words.get(word, word) for word in text.split()])

    # Stemming
    factory = StemmerFactory()
    stemmer = factory.create_stemmer()
    text = stemmer.stem(text)

    # Stop Words Removal
    text = text.split()
    text = " ".join(stopwords_removal(text, sw))

    return text
```

```python
def plot_wordcloud_and_frequency(df, text_column, label_column, stopwords = sw_id, num_word = 10):
    labels = df[label_column].unique()
    num_cols = 2

    for label in labels:
        fig, axes = plt.subplots(1, num_cols, figsize=(12, 6))

        # Filter text by label
        text = " ".join(df[df[label_column] == label][text_column])

        # Generate wordcloud with a colormap
        wordcloud = WordCloud(width=400, height=400,
                              background_color='white',
                              stopwords=stopwords,
                              min_font_size=10,
                              colormap='viridis').generate(text)

        # Display wordcloud
        axes[0].imshow(wordcloud)
        axes[0].axis("off")
        axes[0].set_title(f'WordCloud - {label}')

        # Compute word frequency
        text_data = df[df[label_column] == label][text_column]
        word_frequency = text_data.str.split(expand=True).stack().value_counts()
        freq = word_frequency.head(num_word) # adjust numbers of word

        # Display word frequency bar plot with a color palette
        sns.barplot(x=freq.values, y=freq.index, ax=axes[1], palette="muted")
        axes[1].set_title(f'Word Frequency - {label}')
        axes[1].set_xlabel('Count')
        for p in axes[1].patches:
            axes[1].annotate("{:.0f}".format(p.get_width()), (p.get_x() + p.get_width(), p.get_y() + .5), ha='left', va='center')

        plt.tight_layout()
        plt.show()
```

**Text classification** adalah proses mengkategorikan teks ke dalam beberapa kelas atau kategori. Ini adalah salah satu tugas paling umum dalam pemrosesan bahasa alami (Natural Language Processing atau NLP). Text classification digunakan dalam berbagai aplikasi, termasuk filter spam email, analisis dokumen hukum, dan pengelompokan artikel berita.
<br><br>
Proses text classification umumnya melibatkan beberapa langkah:
1. Pengumpulan Data: Data teks dikumpulkan dari berbagai sumber seperti media sosial, email, artikel berita, dll. Jika belum terdapat label, maka tiap sampel data dapat diberikan label secara manual, misalnya positif, negatif, dan netral pada analisis sentimen.
2. Pra-pemrosesan Data: Teks mentah diproses untuk menghilangkan noise dan menyederhanakan struktur data. Langkah-langkahnya meliputi cleaning data, stop words, stemming, dan lainnya.
3. Feature Extraction: Setelah data teks diproses, tahap selanjutnya adalah ekstraksi fitur. Ini melibatkan mengubah teks menjadi representasi numerik yang dapat dimengerti oleh model machine learning. Beberapa teknik yang umum digunakan adalah Bag of Words, TF-IDF (Term Frequency-Inverse Document Frequency), dan word embeddings seperti Word2Vec atau GloVe.
4. Pemilihan Model: Model machine learning seperti Naive Bayes, Support Vector Machines (SVM), atau neural networks dipilih dan dilatih menggunakan data yang telah direpresentasikan.
5. Evaluasi Model: Model yang telah dilatih dievaluasi menggunakan metrik seperti akurasi, precision, recall, atau F1-score untuk memastikan kinerjanya

### Feature Extraction
Feature extraction adalah proses penting dalam analisis data, terutama dalam konteks pemrosesan bahasa alami (NLP) dan pengenalan pola. Ini melibatkan mengubah data mentah menjadi **representasi numerik** yang lebih sederhana dan informatif yang dapat dimengerti oleh model machine learning. Pada sesi ini akan digunakan **Metode TF-IDF** (Term Frequency-Inverse Document Frequency). TF-IDF adalah teknik yang umum digunakan dalam pemrosesan bahasa alami (NLP) untuk mengekstraksi dan menimbang pentingnya sebuah kata dalam suatu dokumen relatif terhadap semua dokumen dalam kumpulan data.
- Term Frequency (TF) mengukur seberapa sering sebuah kata muncul dalam sebuah dokumen. Semakin sering kata tersebut muncul, semakin besar nilainya.
- Inverse Document Frequency (IDF) mengukur seberapa unik sebuah kata dengan memperhitungkan jumlah dokumen yang mengandung kata tersebut dalam kumpulan data. Kata-kata yang jarang muncul dalam dokumen mendapatkan bobot yang lebih tinggi.
<br><br>
<img src="https://editor.analyticsvidhya.com/uploads/88298tfidf_ex3.png" alt="TF-IDF" width="800" height="400">
<br><br>
Dengan menggabungkan TF dan IDF, TF-IDF memberikan bobot yang lebih besar pada kata-kata yang sering muncul dalam dokumen tertentu tetapi jarang muncul dalam seluruh kumpulan data. Hal ini membantu mengidentifikasi kata-kata yang unik dan penting dalam sebuah dokumen (text), yang dapat digunakan untuk membedakan dokumen tersebut dari yang lain dan untuk mengekstrak informasi yang relevan.

### **Modelling**

```python
# Single Model
def train_model(model, data, targets):
    # Create a Pipeline object with a TfidfVectorizer and the given model
    text_clf = Pipeline([('vect', TfidfVectorizer()),
                         ('clf', model)])

    # Fit the model on the data and targets
    text_clf.fit(data, targets)
    return text_clf

# Model Comparison
def train_model_comparison(models, X_train, y_train, X_val, y_val):
    results = []
    for name, model in models.items():
        text_clf = train_model(model, X_train, y_train)

        # Compute cross-validated scores
        cv_scores = cross_val_score(text_clf, X_train, y_train, cv = kf)
        train_acc = cv_scores.mean()

        # Predict on the validation data
        val_pred = text_clf.predict(X_val)

        # Compute accuracy on validation data
        val_acc = accuracy_score(y_val, val_pred)

        results.append([name, train_acc, val_acc])

    # Create DataFrame for results
    df_results = pd.DataFrame(results, columns=['Model', 'Train Accuracy', 'Val Accuracy'])
    df_results_sorted = df_results.sort_values(by='Val Accuracy', ascending=False)

    return df_results_sorted
```

> #### Baseline Model

```python
# Model Candidate
models = {
    'Random Forest': RandomForestClassifier(random_state = 42),
    'Logistic Regression': LogisticRegression(random_state = 42),
    'Decision Tree': DecisionTreeClassifier(random_state = 42),
    'SVM': SVC(random_state = 42),
    'Naive Bayes': MultinomialNB(),
    # 'XGBoost': XGBClassifier(random_state = 42) # XGBoost perlu menerima label target dalam bentuk numerik, bukan string
    # Add more models as needed
}

results = train_model_comparison(models, X_train, y_train, X_val, y_val)
print(results)
```

3 model yerbaik pada data validasi adalah Logistic Regression, SVM, dan Random Forest. Maka akan dilakukan optimalisasi pada algoritma SVM dan Random Forest menggunakan **hyperparameter tuning**


> #### SVM

```python
# Membuat pipeline dengan TfidfVectorizer dan model SVM
svm_model = Pipeline([('vect', TfidfVectorizer()), ('clf', SVC(probability = True, random_state = 42))])

# Daftar hyperparameter yang ingin dituning
param_grid = {
    'vect__max_df': (0.5, 0.75, 1.0),
    'vect__ngram_range': ((1, 1), (1, 2)),
    'clf__C': (1, 10, 100),
    'clf__kernel': ('linear', 'rbf'),
}

# Inisialisasi GridSearchCV dengan model, parameter grid, dan metrik evaluasi yang diinginkan
svm_grid = GridSearchCV(svm_model, param_grid, cv = kf, n_jobs = -1, verbose = 1, scoring = 'accuracy')

# Melatih GridSearchCV pada data train
svm_grid.fit(X_train, y_train)

# Menampilkan parameter terbaik
print("Best parameters found: ", svm_grid.best_params_)

# Akurasi data train dan validasi
print("Train Accuracy: ", svm_grid.best_score_)

svm_best = svm_grid.best_estimator_
y_pred = svm_best.predict(X_val)
print("Val Accuracy: ", accuracy_score(y_val, y_pred))
```

**Performa SVM hasil tuning lebih baik dibandingkan dengan Logistic Regression**

> #### Random Forest

```python
# Membuat pipeline dengan TfidfVectorizer dan model Random Forest
rf_model = Pipeline([('vect', TfidfVectorizer()), ('clf', RandomForestClassifier(random_state = 42))])

# Daftar hyperparameter yang ingin dituning
param_grid = {
    'vect__max_df': (0.5, 0.75, 1.0),
    'vect__ngram_range': ((1, 1), (1, 2)),
    'clf__n_estimators': [100, 200, 300],
    'clf__max_depth': [None, 10, 20],
}

# Inisialisasi GridSearchCV dengan model, parameter grid, dan metrik evaluasi yang diinginkan
rf_grid = GridSearchCV(rf_model, param_grid, cv = kf, n_jobs = -1, verbose = 1, scoring = 'accuracy')

# Melatih GridSearchCV pada data train
rf_grid.fit(X_train, y_train)

# Menampilkan parameter terbaik
print("Best parameters found: ", rf_grid.best_params_)

# Akurasi data train dan validasi
print("Train Accuracy: ", rf_grid.best_score_)

rf_best = rf_grid.best_estimator_
y_pred = rf_best.predict(X_val)
print("Val Accuracy: ", accuracy_score(y_val, y_pred))
```
